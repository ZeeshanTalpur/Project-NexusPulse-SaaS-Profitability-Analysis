Turns out 2 of the 3 records having same invoice ID but different PaymentID are valid in case of partial payments but the third one is actually just a copy of first one.\
Consequently, any record showing same invoice ID and exact same other attributes like date and time, status ;  is a duplicate and has to be dropped.

```python
payments = payments.drop_duplicates(
    subset=[
        "InvoiceID",
        "PaymentDate",
        "CustomerID",
        "PaymentMethod",
        "Amount",
        "Currency",
        "Status"
    ],
    keep="first"
)
```

```python
# Sum all refund amounts for each invoice
refund_totals = (
    refunds.groupby("InvoiceID")["RefundAmount"]
    .sum()
    .reset_index(name="TotalRefundAmount")
)

# Merge with invoices
refund_check = invoices.merge(
    refund_totals,
    on="InvoiceID",
    how="inner"
)

# Find invoices where refunds exceed the invoice amount
refund_exceeded = refund_check[
    refund_check["TotalRefundAmount"] > refund_check["TotalAmount"]
]
```
