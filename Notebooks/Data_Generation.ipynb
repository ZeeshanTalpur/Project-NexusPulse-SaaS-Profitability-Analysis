```python
def annual_revenue(employees):

    base = employees * np.random.randint(

        10000,

        80000

    )

    noise = np.random.randint(

        -500000,

        500000

    )

    return max(

        100000,

        base + noise

    )
```

```python
for _, sub in subscriptions.iterrows():

    start = sub["StartDate"]

    end = END_DATE

    if pd.notna(sub["CancelDate"]):

        end = min(end, sub["CancelDate"])

    if sub["BillingCycle"] == "Monthly":

        freq = 30

    elif sub["BillingCycle"] == "Quarterly":

        freq = 90

    else:

        freq = 365

    current = start

    while current <= end:

        invoice_records.append({

            "InvoiceID": f"I{invoice_number:07}",

            "SubscriptionID": sub["SubscriptionID"],

            "CustomerID": sub["CustomerID"],

            "InvoiceDate": current,

            "DueDate": current + pd.Timedelta(days=30),

            "PlanID": sub["PlanID"],

            "PlanName": sub["PlanName"],

            "MonthlyPrice": sub["MonthlyPrice"],

            "DiscountPercent": sub["DiscountPercent"],

            "Status": np.random.choice(

                [

                    "Paid",

                    "Pending",

                    "Overdue"

                ],

                p=[

                    0.82,

                    0.10,

                    0.08

                ]

            )

        })

        invoice_number += 1

        current += pd.Timedelta(days=freq)
```

```python
payment_records = []

payment_number = 1
```

```python
payment_methods = [

    "Credit Card",

    "Debit Card",

    "Bank Transfer",

    "PayPal",

    "Stripe",

    "Apple Pay",

    "Google Pay"

]
```

```python
for _, invoice in invoices.iterrows():

    status = invoice["Status"]

    total = invoice["TotalAmount"]

    if status == "Pending":

        continue

    if status == "Overdue":

        if np.random.random() < 0.75:

            continue

    payment_count = np.random.choice(

        [1,2],

        p=[0.93,0.07]

    )

    remaining = total

    for i in range(payment_count):

        if payment_count == 1:

            amount = total

        else:

            if i == 0:

                amount = round(

                    total *

                    np.random.uniform(0.3,0.7),

                    2

                )

                remaining -= amount

            else:

                amount = round(

                    remaining,

                    2

                )

        payment_records.append({

            "PaymentID":f"P{payment_number:07}",

            "InvoiceID":invoice["InvoiceID"],

            "CustomerID":invoice["CustomerID"],

            "PaymentDate":invoice["InvoiceDate"] +

                pd.Timedelta(

                    days=np.random.randint(

                        0,

                        45

                    )

                ),

            "PaymentMethod":np.random.choice(

                payment_methods

            ),

            "Amount":amount,

            "Currency":invoice["Currency"],

            "Status":"Completed"

        })

        payment_number += 1
```

```python
support_agents = employees.loc[

    employees["Department"]=="Support",

    "EmployeeID"

].values
```

```python
for _, customer in customers.iterrows():

    ticket_count = np.random.poisson(2)

    ticket_count = min(ticket_count,8)

    for _ in range(ticket_count):

        created = random_dates(

            customer["JoinDate"],

            END_DATE,

            1

        )[0]

        resolution = created + pd.Timedelta(

            hours=np.random.randint(

                1,

                240

            )

        )

        satisfaction = np.random.randint(

            1,

            6

        )

        ticket_records.append({

            "TicketID":f"T{ticket_number:07}",

            "CustomerID":customer["CustomerID"],

            "AgentID":np.random.choice(

                support_agents

            ),

            "Category":np.random.choice(

                categories

            ),

            "Priority":np.random.choice(

                priorities,

                p=[

                    0.35,

                    0.40,

                    0.20,

                    0.05

                ]

            ),

            "Status":np.random.choice(

                statuses,

                p=[

                    0.55,

                    0.20,

                    0.15,

                    0.10

                ]

            ),

            "CreatedDate":created,

            "ResolvedDate":resolution,

            "ResolutionHours":round(

                (

                    resolution-created

                ).total_seconds()/3600,

                2

            ),

            "CustomerSatisfaction":satisfaction

        })

        ticket_number += 1
```

```python
support_tickets = pd.DataFrame(

    ticket_records

)
```
